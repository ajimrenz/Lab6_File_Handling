```python
# CELL 1 - STUDENT CONFIGURATION

from pathlib import Path

LAST_NAME = "Agawa"
STUDENT_ID = "TUPM-26-1082"

SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())
NAME_LENGTH = len(LAST_NAME)

print("Last Name:", LAST_NAME)
print("Student ID:", STUDENT_ID)
print("Seed:", SEED_DIGIT)
print("ID Sum:", ID_SUM)
print("Name Length:", NAME_LENGTH)
```

```python
# CELL 2 - CREATE ACTIVITY WORKSPACE

WORK_DIR = Path.cwd()
DATA_DIR = WORK_DIR / "data"
OUTPUT_DIR = WORK_DIR / "output"
BACKUP_DIR = WORK_DIR / "backup"
for folder in (DATA_DIR, OUTPUT_DIR, BACKUP_DIR):
    folder.mkdir(parents=True, exist_ok=True)
print("Workspace:", WORK_DIR)
for folder in (DATA_DIR, OUTPUT_DIR, BACKUP_DIR):
    print(folder.name, "exists:", folder.exists())
```

```python
# CELL 3 - PATH COMPONENTS AND TEXT FILE

file_path = WORK_DIR / f"{LAST_NAME}_{STUDENT_ID}_notes.txt"
print("Full path:", file_path)
print("File name:", file_path.name)
print("Extension:", file_path.suffix)
print("Parent folder:", file_path.parent)
with open(file_path, "w", encoding="utf-8") as file:
    file.write(f"Student: {LAST_NAME}\n")
    file.write(f"Student ID: {STUDENT_ID}\n")
    file.write(f"Seed digit: {SEED_DIGIT}\n")
    file.write(f"ID sum: {ID_SUM}\n")
print("Created:", file_path.exists())
```

```python
# CELL 4 - APPEND VERIFICATION

verification_value = ID_SUM * NAME_LENGTH + SEED_DIGIT
with open(file_path, "a", encoding="utf-8") as file:
    file.write(f"Verification value: {verification_value}\n")
with open(file_path, "r", encoding="utf-8") as file:
    print(file.read())
```

```python
# CELL 5 - READ METHODS

with open(file_path, "r", encoding="utf-8") as file:
    print("read():\n", file.read())
with open(file_path, "r", encoding="utf-8") as file:
    print("readline():", file.readline())
with open(file_path, "r", encoding="utf-8") as file:
    print("readlines():", file.readlines())
```

```python
# CELL 6 - READ LINE BY LINE

with open(file_path, "r", encoding="utf-8") as file:
    for line_number, line in enumerate(file, start=1):
        print(f"Line {line_number}: {line.strip()}")
```

```python
# CELL 7 - FILTER LINES

filtered_path = OUTPUT_DIR / f"{LAST_NAME}_{STUDENT_ID}_filtered.txt"
with open(file_path, "r", encoding="utf-8") as source, open(filtered_path, "w", encoding="utf-8") as output:
    for line in source:
        if "ID" in line or "Verification" in line:
            output.write(line)
print("Filtered file:", filtered_path)
print(filtered_path.read_text(encoding="utf-8"))
```

```python
# CELL 8 - CREATE AND PARSE CSV

import csv
csv_path = OUTPUT_DIR / f"{LAST_NAME}_{STUDENT_ID}_record.csv"
with open(csv_path, "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerow(["last_name", "student_id", "seed_digit", "id_sum"])
    writer.writerow([LAST_NAME, STUDENT_ID, SEED_DIGIT, ID_SUM])
with open(csv_path, "r", newline="", encoding="utf-8") as file:
    rows = list(csv.reader(file))
print("Fields:", rows[0])
print("Values:", rows[1])
```

```python
# CELL 9 - WRITE AND VERIFY JSON

import json
json_path = OUTPUT_DIR / f"{LAST_NAME}_{STUDENT_ID}_record.json"
student_data = {"last_name": LAST_NAME, "student_id": STUDENT_ID, "seed_digit": SEED_DIGIT, "id_sum": ID_SUM, "name_length": NAME_LENGTH, "verification_value": verification_value}
with open(json_path, "w", encoding="utf-8") as file:
    json.dump(student_data, file, indent=4)
with open(json_path, "r", encoding="utf-8") as file:
    loaded = json.load(file)
expected = loaded["id_sum"] * loaded["name_length"] + loaded["seed_digit"]
print("Stored verification:", loaded["verification_value"])
print("Expected verification:", expected)
print("Status:", "VERIFIED" if loaded["verification_value"] == expected else "CHECK")
```

```python
# CELL 10 - FILE METADATA AND BACKUP

import shutil
file_info = csv_path.stat()
print("File name:", csv_path.name)
print("File size:", file_info.st_size, "bytes")
print("Extension:", csv_path.suffix)
backup_path = BACKUP_DIR / csv_path.name
shutil.copy2(csv_path, backup_path)
print("Backup exists:", backup_path.exists())
```

```python
# CELL 11 - READ HOUSEHOLD POWER DATASET

dataset_path = DATA_DIR / "household_power_consumption.txt"
print("Dataset path:", dataset_path.resolve())
if not dataset_path.is_file():
    raise FileNotFoundError(f"Place household_power_consumption.txt in {DATA_DIR}")
with open(dataset_path, "r", encoding="utf-8") as file:
    header = file.readline().strip()
    records = [file.readline().strip() for _ in range(3)]
print("Header:", header)
print("Sample records:")
for record in records:
    print(record)
```

```python
# CELL 12 - PROCESS DATASET SAMPLE

sample_size = 50 + ID_SUM
total_power = 0.0
valid_records = 0
with open(dataset_path, "r", encoding="utf-8") as file:
    next(file)
    for _ in range(sample_size):
        line = file.readline()
        if not line:
            break
        fields = line.strip().split(";")
        if len(fields) > 2 and fields[2] != "?":
            try:
                total_power += float(fields[2])
                valid_records += 1
            except ValueError:
                pass
average_power = total_power / valid_records if valid_records else 0.0
print("Sample size:", sample_size)
print("Valid records:", valid_records)
print("Average Global Active Power:", round(average_power, 3))
```

```python
# CELL 13 - GENERATE PROCESSING REPORT

report_path = OUTPUT_DIR / f"{LAST_NAME}_{STUDENT_ID}_processing_report.txt"
report_lines = [
    "PYTHON FILE-HANDLING PROCESSING REPORT",
    "--------------------------------------",
    f"Last Name: {LAST_NAME}",
    f"Student ID: {STUDENT_ID}",
    f"Seed Digit: {SEED_DIGIT}",
    f"ID Sum: {ID_SUM}",
    f"Verification Value: {verification_value}",
    f"Dataset Sample Size: {sample_size}",
    f"Valid Records: {valid_records}",
    f"Average Global Active Power: {average_power:.3f}",
    f"JSON File: {json_path.name}",
    f"Backup File: {backup_path.name}",
]
report_path.write_text("\n".join(report_lines) + "\n", encoding="utf-8")
print(report_path.read_text(encoding="utf-8"))
```
